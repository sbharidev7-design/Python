```python
marks = {
    "anu":85,
    "rahul":72,
    "akhil":91,
    "meera":65,
    "neha":78
}
```

```python
#student with highest mark
highestmark=max(marks,key=marks.get)
print('student with highest mark:',highestmark)
```

```python
#student with low mark
lowestmark=min(marks,key=marks.get)
print('student with lowest mark:',lowestmark)
```

```python
#average marks of students
totalmarks=sum(marks.values())
avg=totalmarks/len(marks)
print('average mark',avg)
```

```python
#create a new dictnonary
new={}
for name,mark in marks.items():
 if mark>=75:
    new[name]=mark
    print(new)
```

```python
#pattern printing
for i in range(6):
    for j in range(1,i+1):
         print(j,end='')
    print()
```

```python
#reverse 
for i in range(5, 0, -1):
    for j in range(1,i+1):
         print(j,end='')
    print()
```

```python
balance = 10000

amount=int(input("enter withdrawal amount: "))

if amount <=0:
    print("withdrawal amount invalid.")
elif amount > balance:
    print("balance insufficent")
else:
    balance = balance - amount
    print("withdrawal successful.")
    print("remaining balance:" , balance)
```
